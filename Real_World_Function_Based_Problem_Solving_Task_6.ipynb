```python
#Function to manage smart parking system
def manage_parking(capacity,logs):
  parked=0
  peak=0
  #Loop through vehicle logs to track entry and exit
  for log in logs:
    if log=="IN":
      parked=parked+1
    elif log=="OUT":
      parked=parked-1
    #Tracking peak usage
    if parked>peak:
      peak=parked
  #Checking parking status
  if parked>capacity:
    status="Capacity Exceeded"
  else:
    status="Available"
  print("Currently Parked Vehicles:",parked)
  print("Parking Status:",status)

capacity=50
logs=["IN","IN","IN","OUT","IN","IN","OUT"]
manage_parking(capacity,logs)
```

```python
#Function to estimate delivery time
def estimate_delivery(distance,traffic,weather):
  #Base time assuming 5 minutes per km
  time=distance*5
  #Adding delay based on traffic
  if traffic=="High":
    time=time+15
  elif traffic=="Medium":
    time=time+10
  #Adding delay based on weather
  if weather=="Rainy":
    time=time+10
  elif weather=="Storm":
    time=time+20
  print("Estimated Delivery Time:",time,"minutes")

distance=8
traffic="High"
weather="Rainy"
estimate_delivery(distance,traffic,weather)
```

```python
#Function to analyze seat occupancy
def analyze_occupancy(total_seats,booked_seats):
  booked=0
  #Loop to count booked seats
  for seat in booked_seats:
    if seat==1:
      booked=booked+1
  occupancy=(booked/total_seats)*100
  #Checking show status
  if occupancy==100:
    status="Housefull"
  elif occupancy>=75:
    status="Almost Full"
  else:
    status="Seats Available"
  print("Occupancy:",occupancy,"%")
  print("Show Status:",status)

total_seats=200
booked_seats=[1]*150
analyze_occupancy(total_seats,booked_seats)
```

```python
#Function to classify server load
def classify_server(cpu_readings):
  total=0
  #Loop to calculate total CPU usage
  for cpu in cpu_readings:
    total=total+cpu
  average=total/len(cpu_readings)
  #Checking load condition
  if average<50:
    status="Normal"
  elif average<=80:
    status="Warning"
  else:
    status="Critical"
  print("Average CPU Load:",average,"%")
  print("Server Status:",status)

cpu_readings=[45,60,70,85,90]
classify_server(cpu_readings)
```

```python
#Function to monitor classroom resource usage
def monitor_resources(resources):
  overused=[]
  #Loop through resource usage dictionary
  for resource in resources:
    hours=resources[resource]
    #Checking if usage is above threshold (8 hours)
    if hours>8:
      overused.append(resource)
  if len(overused)>0:
    alert="Yes"
  else:
    alert="No"
  print("Overused Resources:",",".join(overused))
  print("Energy Alert:",alert)

resources={
 "Projector":6,
 "AC":9,
 "Lights":4
}
monitor_resources(resources)
```

```python
#Function to manage event registrations
def manage_registration(capacity,registrations):
  confirmed=0
  waitlist=0
  #Checking registration count against capacity
  if registrations<=capacity:
    confirmed=registrations
  else:
    confirmed=capacity
    waitlist=registrations-capacity
  #Checking registration status
  if registrations>capacity:
    status="Closed"
  else:
    status="Open"
  print("Confirmed Registrations:",confirmed)
  print("Waitlisted Users:",waitlist)
  print("Registration Status:",status)
capacity=100
registrations=105
manage_registration(capacity,registrations)
```
