```python
#Class to control smart light
class SmartLight:
  #Initializing light name and default status
  def __init__(self,name):
    self.name=name
    self.status="OFF"
  #Method to turn light ON
  def turn_on(self):
    self.status="ON"
  #Method to turn light OFF
  def turn_off(self):
    self.status="OFF"
  #Method to display current status
  def display_status(self):
    print(self.name,"is",self.status)
light=SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

```python
#Class to store employee ID card details
class Employee:
  #Initializing employee details
  def __init__(self,name,emp_id,department):
    self.name=name
    self.emp_id=emp_id
    self.department=department
  #Method to display ID card
  def display_id_card(self):
    print("Employee ID Card")
    print("Name:",self.name)
    print("ID:",self.emp_id)
    print("Department:",self.department)
emp=Employee("Rahul","EMP102","IT")
emp.display_id_card()
```

```python
#Class to store contact information
class Contact:
  #Initializing contact details
  def __init__(self,name,phone):
    self.name=name
    self.phone=phone
  #Method to display contact info
  def display_contact(self):
    print("Contact Saved")
    print("Name:",self.name)
    print("Phone:",self.phone)

contact=Contact("Anita","9876543210")
contact.display_contact()
```

```python
#Class to generate product price tag
class Product:
  #Initializing product name and price
  def __init__(self,name,price):
    self.name=name
    self.price=price
  #Method to display price tag
  def display_price(self):
    print("Product:",self.name)
    print("Price: ₹"+str(self.price))

product=Product("Headphones",2499)
product.display_price()
```

```python
#Class to store movie rating details
class Movie:
  #Initializing movie name and rating
  def __init__(self,name,rating):
    self.name=name
    self.rating=rating
  #Method to display movie info
  def display_movie(self):
    print("Movie:",self.name)
    print("Rating:",self.rating,"/ 5")

movie=Movie("Inception",4.8)
movie.display_movie()
```

```python
#Class to manage delivery details
class Delivery:
  #Initializing customer name and address
  def __init__(self,name,address):
    self.name=name
    self.address=address
  #Method to display delivery details
  def display_delivery(self):
    print("Delivery Details")
    print("Customer:",self.name)
    print("Address:",self.address)
delivery=Delivery("Suman","Hyderabad")
delivery.display_delivery()
```
